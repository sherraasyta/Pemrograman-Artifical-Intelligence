```python
#Gamar 5.1
def kalimat():                                  # deklarasi fungsi
    """menampilkan kalimat Hallo, Selamat Belajar Python"""   # docstring
    print("Hallo, Selamat Belajar Python")       # isi fungsi

kalimat()                                       # memanggil fungsi
```

```python
#Gambar 5.2
def penjumlahan():
    """menghitung penjumlahan 1 dengan 2"""
    hasil = 1 + 2                               # proses hitung
    return hasil                                # kembalikan nilai ke pemanggil

tampil = penjumlahan()                          # tampung nilai kembalian
print("Hasil penjumlahan 1 + 2 =", tampil)
```

```python
#Gambar 5.3
def identitas(nama, kota):                      # dua parameter
    print("Nama           :", nama)
    print("Kota kelahiran :", kota)

nama = input("Masukkan nama: ")                 # input dari pengguna
kota = input("Masukkan kota kelahiran: ")
identitas(nama, kota)                           # kirim argumen ke fungsi
```

```python
#Gambar 5.4 dan 5.5: fungsi dengan parameter dan return, kata kunci
def perkalian(bil1, bil2):
    hasil = bil1 * bil2                         # kalikan kedua parameter
    return hasil

print(perkalian(5, 7))                          # urutan argumen biasa -> 35
print(perkalian(bil2=7, bil1=5))                # pakai kata kunci -> 35
```

```python
#Gambar 5.6 dan 5.7: variabel lokal dan global
x_global = 10                                   # variabel global

def variabel_lokal():
    y_lokal = 5                                 # variabel lokal
    print("x global =", x_global)               # global bisa dibaca di fungsi
    print("y lokal  =", y_lokal)

variabel_lokal()
print("x global di luar =", x_global)
try:
    print(y_lokal)                              # lokal tidak dikenal di luar fungsi
except NameError:
    print("y_lokal tidak dikenal (NameError)")
```

```python
#Pertemuan 4: bilangan dengan pencacah 3 dan hasil perkalian
a = int(input("Bilangan awal : "))
b = int(input("Bilangan akhir: "))
hasil = 1                                       # nilai awal perkalian
for i in range(a, b + 1, 3):                    # loncat 3 (pencacah 3)
    print(i, end=" ")
    hasil *= i                                  # kalikan semua bilangan
print("\nHasil perkalian =", hasil)
```

```python
#Pertemuan 4: penjumlahan 2 matriks
baris = int(input("Jumlah baris : "))
kolom = int(input("Jumlah kolom : "))
A, B = [], []
for i in range(baris):                          # isi matriks A
    A.append([int(input("A[%d][%d] = " % (i, j))) for j in range(kolom)])
for i in range(baris):                          # isi matriks B
    B.append([int(input("B[%d][%d] = " % (i, j))) for j in range(kolom)])
for i in range(baris):                          # tampilkan A + B
    for j in range(kolom):
        print("%4d" % (A[i][j] + B[i][j]), end="")
    print()                                     # pindah baris
```

```python
#Pertemuan 4, Soal 1: total pembelian N barang
n = int(input("Banyak barang: "))
total = 0
for i in range(1, n + 1):
    nama = input("Nama barang ke-%d : " % i)
    harga = int(input("  Harga satuan   : "))
    jumlah = int(input("  Jumlah beli    : "))
    total += harga * jumlah                     # tambah subtotal
print("Total pembelian = Rp", total)
```

```python
#2
n = int(input("Banyak bilangan: "))
genap = ganjil = 0                              # dua penghitung
for i in range(1, n + 1):
    angka = int(input("Bilangan ke-%d: " % i))
    if angka % 2 == 0:                          # sisa bagi 2 = 0 -> genap
        genap += 1
    else:
        ganjil += 1
print("Genap =", genap, "| Ganjil =", ganjil)
```

```python
#5
n = int(input("Banyak mahasiswa: "))
data = []
for i in range(n):
    nama = input("Nama    : ")
    tugas = float(input("N.Tugas : "))
    kuis = float(input("N.Kuis  : "))
    uts = float(input("N.UTS   : "))
    uas = float(input("N.UAS   : "))
    akhir = 0.20*tugas + 0.20*kuis + 0.25*uts + 0.35*uas   # bobot asumsi
    data.append((nama, tugas, kuis, uts, uas, akhir))
print("%-4s %-12s %8s %8s %8s %8s %11s" %
      ("No.", "Nama Mhs", "N.Tugas", "N.Kuis", "N.UTS", "N.UAS", "NilaiAkhir"))
for no, d in enumerate(data, 1):
    print("%-4d %-12s %8.1f %8.1f %8.1f %8.1f %11.2f" % ((no,) + d))
```
