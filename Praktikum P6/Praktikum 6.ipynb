```python
#Contoh penggunaan Nested Loop

i = 2
while(i < 100):
    j = 2
    while(j <= (i/j)):
        if not(i%j): break
        j = j + 1
    if (j > i/j) : print(i, " is prime")
    i = i + 1

print("Good bye!")
```

```python
# Program 1: Memfilter dan menghitung jumlah bilangan genap & ganjil dari suatu list

list_angka = [1, 2, 3, 4, 5, 6, 7, 8, 9]

jumlah_genap = 0
jumlah_ganjil = 0

for angka in list_angka:
    if angka % 2 == 0:
        jumlah_genap += 1
    else:
        jumlah_ganjil += 1

print("List angka:", tuple(list_angka))
print("Jumlah bilangan genap:", jumlah_genap)
print("Jumlah bilangan ganjil:", jumlah_ganjil)
```

```python
# Program 2: Menampilkan nilai perkalian dari masukan nilai

input_angka = int(input("Input sebuah angka: "))

for i in range(1, 11):
    hasil = input_angka * i
    print(f"{input_angka} x {i} = {hasil}")
```

```python
# Program 3: Menampilkan urutan nilai Fibonacci

n = int(input("Masukkan jumlah suku Fibonacci: "))

a, b = 0, 1
count = 0

print("Deret Fibonacci:")
while count < n:
    print(a, end=" ")
    a, b = b, a + b
    count += 1
print()
```

```python
# Program 4: Perkalian nilai 1 sampai 100 (Tabel Perkalian 1 sampai 10)

for i in range(1, 11):
    for j in range(1, 11):
        print(f"{i * j:4}", end="")
    print()
```
