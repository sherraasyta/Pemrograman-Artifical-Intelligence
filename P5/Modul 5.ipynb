```python
#Kondisi if adalah kondisi yang akan dieksekusi oleh program jika bernilai benar atau TRUE

nilai = 10

#jika kondisi benar/TRUE maka program akan mengeksekusi perintah dibawahnya
if(nilai > 7):
    print("Selamat Anda Lulus")

#jika kondisi salah/FALSE maka program tidak akan mengeksekusi perintah dibawahnya
if(nilai > 10):
    print("Selamat Anda Lulus")
```

```python
#Kondisi if else adalah jika kondisi bernilai TRUE maka akan dieksekusi pada if, tetapi jika bernilai FALSE maka akan dieksekusi kode pada else

nilai = 7
#Jika pernyataan pada if bernilai TRUE maka if akan dieksekusi, tetapi jika FALSE kode pada else yang akan dieksekusi.
if(nilai > 7):
    print("Selamat Anda Lulus")
else:
    print("Maaf Anda Tidak Lulus")
```

```python
#LATIHAN 
#A
print("--- Program Pengecekan Jenis Segitiga ---")
x = int(input("Masukkan nilai x: "))
y = int(input("Masukkan nilai y: "))
z = int(input("Masukkan nilai z: "))

if x == y == z:
    print("Keputusan: Segitiga Sama Sisi (Equilateral Triangle)")
elif x == y or x == z or y == z:
    print("Keputusan: Segitiga Sama Kaki (Isosceles Triangle)")
else:
    print("Keputusan: Segitiga Sembarang (Scalene Triangle)")
```

```python
#B
print("--- Program Pengecekan Tahun Kabisat ---")
tahun = int(input("Masukkan tahun: "))

# Tahun kabisat adalah tahun yang habis dibagi 400, atau habis dibagi 4 tapi tidak habis dibagi 100
if (tahun % 400 == 0) or (tahun % 4 == 0 and tahun % 100 != 0):
    print(f"Tahun {tahun} merupakan Tahun Kabisat")
else:
    print(f"Tahun {tahun} bukan merupakan Tahun Kabisat")
```

```python
#C
print("--- Program Deteksi Zodiak Yunani ---")
tanggal = int(input("Tanggal lahir anda: "))
bulan = input("Bulan anda lahir (contoh: januari): ").lower()

if (bulan == 'desember' and tanggal >= 22) or (bulan == 'januari' and tanggal <= 19):
    zodiak = "Capricorn"
elif (bulan == 'januari' and tanggal >= 20) or (bulan == 'februari' and tanggal <= 18):
    zodiak = "Aquarius"
elif (bulan == 'februari' and tanggal >= 19) or (bulan == 'maret' and tanggal <= 20):
    zodiak = "Pisces"
elif (bulan == 'maret' and tanggal >= 21) or (bulan == 'april' and tanggal <= 19):
    zodiak = "Aries"
elif (bulan == 'april' and tanggal >= 20) or (bulan == 'mei' and tanggal <= 20):
    zodiak = "Taurus"
elif (bulan == 'mei' and tanggal >= 21) or (bulan == 'juni' and tanggal <= 20):
    zodiak = "Gemini"
elif (bulan == 'juni' and tanggal >= 21) or (bulan == 'juli' and tanggal <= 22):
    zodiak = "Cancer"
elif (bulan == 'juli' and tanggal >= 23) or (bulan == 'agustus' and tanggal <= 22):
    zodiak = "Leo"
elif (bulan == 'agustus' and tanggal >= 23) or (bulan == 'september' and tanggal <= 22):
    zodiak = "Virgo"
elif (bulan == 'september' and tanggal >= 23) or (bulan == 'oktober' and tanggal <= 22):
    zodiak = "Libra"
elif (bulan == 'oktober' and tanggal >= 23) or (bulan == 'november' and tanggal <= 21):
    zodiak = "Scorpio"
elif (bulan == 'november' and tanggal >= 22) or (bulan == 'desember' and tanggal <= 21):
    zodiak = "Sagittarius"
else:
    zodiak = "Tidak valid"

print(f"Astrologi anda adalah {zodiak}")
```

```python
# D PROGRAM 4
print("--- Program Deteksi Zodiak Cina ---")
tahun_lahir = int(input("Tahun lahir anda : "))

# Menggunakan modulo 12 dari tahun untuk menentukan shio (Zodiak Cina)
sisa = tahun_lahir % 12

if sisa == 0:
    shio = "Monyet"
elif sisa == 1:
    shio = "Ayam"
elif sisa == 2:
    shio = "Anjing"
elif sisa == 3:
    shio = "Babi"
elif sisa == 4:
    shio = "Tikus"
elif sisa == 5:
    shio = "Kerbau"
elif sisa == 6:
    shio = "Macan"
elif sisa == 7:
    shio = "Kelinci"
elif sisa == 8:
    shio = "Naga"
elif sisa == 9:
    shio = "Ular"
elif sisa == 10:
    shio = "Kuda"
elif sisa == 11:
    shio = "Kambing"
else:
    shio = "Tidak terdeteksi"

print(f"Tahun {tahun_lahir}, zodiak cina nya adalah {shio}")
```
