```python
# Kondisi if
nilai = 9
#jika kondisi benar/TRUE maka program akan mengeksekusi perintah dibawahnya
if(nilai > 7):
    print("Sembilan Lebih Besar Dari Angka Tujuh") # Kondisi Benar, Dieksekusi
#jika kondisi salah/FALSE maka program tidak akan mengeksekusi perintah dibawahnya
if(nilai > 10):
    print("Sembilan Lebih Besar Dari Angka Sepuluh") # Kondisi Salah, Maka tidak tereksekusi
```

```python
#Kondisi if else adalah jika kondisi bernilai TRUE maka akan dieksekusi pada if, tetapi jika bernilai FALSE maka akan dieksekusi kode pada else
nilai = 9
#Jika pernyataan pada if bernilai TRUE maka if akan dieksekusi, tetapi jika FALSE kode pada else yang akan dieksekusi.
if(nilai > 7):
    print("Selamat Anda Lulus")
else:
    print("Maaf Anda Tidak Lulus")
```

```python
# Tarif per bulan untuk buka blokir
biaya_per_bulan = 800000
daftar_bulan = ["JANUARI", "FEBRUARI", "MARET", "APRIL", "MEI", "JUNI", "JULI"]

# Representasi data (B = BLOKIR, U = BUKA BLOKIR, - = Kosong)
data_mahasiswa = {
    "A=Lia"   : ["U", "B", "B", "U", "-", "-", "-"],
    "B=Angga" : ["B", "B", "B", "U", "-", "-", "-"],
    "C=Dimas" : ["B", "B", "B", "B", "U", "-", "-"],
    "D=Syakir": ["B", "B", "B", "B", "B", "U", "-"],
    "E=Irham" : ["B", "B", "B", "B", "B", "B", "U"],
    "F=Zilda" : ["U", "U", "B", "B", "B", "B", "U"],
    "G=Lana"  : ["B", "U", "B", "B", "U", "B", "U"]
}

# Mencetak Header Tabel
header = f"{'Nama Mahasis':<12} | {'JANUARI':<11} | {'FEBRUARI':<11} | {'MARET':<11} | {'APRIL':<11} | {'MEI':<11} | {'JUNI':<11} | {'JULI':<11}"
print(header)
print("-" * len(header))

# Memproses dan mencetak data tiap mahasiswa
for nama, status_bulanan in data_mahasiswa.items():
    akumulasi_bulan = 0
    row_data = [f"{nama:<12}"]
    
    for i in range(len(status_bulanan)):
        status = status_bulanan[i]
        
        if status == "B":
            akumulasi_bulan += 1
            row_data.append(f"{'BLOKIR':<11}")
            
        elif status == "U":
            akumulasi_bulan += 1
            total_bayar = akumulasi_bulan * biaya_per_bulan
            
            # Jika total hanya 1 bulan (800rb), tulis "BUKA BLOKIR" sesuai tabel asli
            if total_bayar == 800000:
                teks_cetak = "BUKA BLOKIR"
            else:
                # Menampilkan nominal hasil perhitungan (termasuk untuk G=Lana)
                teks_cetak = str(total_bayar)
                
            row_data.append(f"{teks_cetak:<11}")
            akumulasi_bulan = 0 # Reset akumulasi
            
        else:
            row_data.append(f"{'':<11}")
            
    # Menggabungkan seluruh kolom menjadi 1 baris
    print(" | ".join(row_data))
```

```python
# Tarif per bulan untuk buka blokir
biaya_per_bulan = 800000

# Data status bulan Januari - Juli untuk G=Lana
status_lana = [
    "BLOKIR",       # Januari
    "BUKA BLOKIR",  # Februari
    "BLOKIR",       # Maret
    "BLOKIR",       # April
    "BUKA BLOKIR",  # Mei
    "BLOKIR",       # Juni
    "BUKA BLOKIR"   # Juli
]

nama_bulan = ["Januari", "Februari", "Maret", "April", "Mei", "Juni", "Juli"]

# Variabel untuk menyimpan akumulasi bulan yang belum dibayar
akumulasi_bulan = 0

print("Rincian Pembayaran BUKA BLOKIR untuk G=Lana:")
print("-" * 45)

for i in range(len(status_lana)):
    # Setiap bulan berjalan akan menambah akumulasi tagihan 1 bulan
    akumulasi_bulan += 1 
    
    if status_lana[i] == "BUKA BLOKIR":
        # Hitung total biaya yang harus dibayar
        total_bayar = akumulasi_bulan * biaya_per_bulan
        
        # Format angka menjadi format Rupiah
        format_rupiah = f"Rp {total_bayar:,.0f}".replace(",", ".")
        
        print(f"Bulan {nama_bulan[i]:<10}: BUKA BLOKIR = {format_rupiah}")
        
        # Reset akumulasi bulan karena sudah dibayar
        akumulasi_bulan = 0 
    else:
        print(f"Bulan {nama_bulan[i]:<10}: BLOKIR")
```
